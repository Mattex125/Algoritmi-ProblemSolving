# ES2A vs ES2B Benchmark Comparison
Compare performance of es2A.c vs es2B.c with varying problem sizes

```python
import matplotlib.pyplot as plt
import subprocess

# Compile es2A and es2B
print("Compiling es2A.c and es2B.c...")
subprocess.run(["gcc", "-O2", "-w", "es2A.c", "-o", "es2A"])
subprocess.run(["gcc", "-O2", "-w", "es2B.c", "-o", "es2B"])
print("Compilation complete!")
```

```python
import time
import statistics
import subprocess

def gen_test(T, A, B, C):
    return f"{T}\n{A} {B} {C}"

def measure_once(exe, data):
    t0 = time.perf_counter()
    p = subprocess.Popen(
        [exe],
        stdin=subprocess.PIPE,
        stdout=subprocess.DEVNULL,
        stderr=subprocess.DEVNULL,
        text=True,
    )
    p.communicate(data)
    return time.perf_counter() - t0

def measure(exe, data, repeats=3, warmup=1):
    for _ in range(warmup):
        measure_once(exe, data)

    samples = [measure_once(exe, data) for _ in range(repeats)]
    return statistics.median(samples)

# Recursive es2A explodes quickly: keep T small for fair/finishable comparison.
sizes4test = list(range(5, 112, 5))
A, B, C = 3, 5, 7

datasets = {T: gen_test(T, A, B, C) for T in sizes4test}
timesA, timesB = [], []

print(f"Running benchmarks with A={A}, B={B}, C={C}...")
for T in sizes4test:
    data = datasets[T]
    tA = measure("./es2A", data)
    tB = measure("./es2B", data)

    timesA.append(tA)
    timesB.append(tB)

    print(f"T={T:3d}: es2A={tA*1000:7.3f}ms | es2B={tB*1000:7.3f}ms")

print("Benchmarks complete!")
```

```python
import matplotlib.pyplot as plt

# Plot comparison
#plt.figure(figsize=(10, 6))
plt.plot(sizes4test, timesA, label="es2A", color="red")
plt.plot(sizes4test, timesB, label="es2B", color="blue")
plt.xlabel("Input size (T)")
plt.ylabel("Time execution (seconds)")
plt.title("ES2A vs ES2B Performance Comparison")
plt.ylim(0, max(max(timesA), max(timesB)) )
plt.legend()
plt.grid()
plt.show()
```
